```python
# Carregamento dos dados e limpeza inicial
import pandas as pd
from scipy import stats
import numpy as np

# Carregar o dataset
df = pd.read_csv('./data/tested.csv')

# Remover valores nulos de Fare (apenas 1 caso)
df = df.dropna(subset=['Fare'])

print(f"Total de passageiros: {len(df)}")
print(f"Colunas: {df.columns.tolist()}")
print(f"\nTarifa média geral: {df['Fare'].mean():.4f}")
print(f"Desvio padrão da tarifa: {df['Fare'].std():.4f}")
print(f"\nDistribuição por sexo:\n{df['Sex'].value_counts()}")
print(f"\nDistribuição por classe:\n{df['Pclass'].value_counts()}")
print(f"\nDistribuição por sobrevivência:\n{df['Survived'].value_counts()}")
```

```python
# EXERCÍCIO 1: Teste t para uma amostra
# Pergunta: A tarifa média é diferente de 30?
# H0: mu = 30  (a tarifa média populacional é 30)
# H1: mu != 30  (a tarifa média é diferente de 30) — teste bilateral

alpha = 0.05
mu0 = 30

tarifas = df['Fare']
t_stat, p_value = stats.ttest_1samp(tarifas, mu0)

print(f"Tarifa média amostral: {tarifas.mean():.4f}")
print(f"Desvio padrão amostral: {tarifas.std():.4f}")
print(f"n = {len(tarifas)}")
print(f"Estatística t: {t_stat:.4f}")
print(f"p-valor: {p_value:.6f}")
print()

if p_value < alpha:
    print(f"p-valor ({p_value:.6f}) < alpha ({alpha}) -> Rejeitamos H0")
    print("Conclusão: A tarifa média é significativamente diferente de 30.")
else:
    print(f"p-valor ({p_value:.6f}) >= alpha ({alpha}) -> Não rejeitamos H0")
    print("Conclusão: Não há evidência de que a tarifa média seja diferente de 30.")
```

```python
# EXERCÍCIO 2: Teste t para DUAS amostras independentes
# Pergunta: Homens e mulheres pagaram tarifas médias diferentes?
# H0: mu_homens = mu_mulheres
# H1: mu_homens != mu_mulheres — teste bilateral
# Usamos o teste t de Welch (variâncias desiguais)

alpha = 0.05

fare_male = df[df['Sex'] == 'male']['Fare']
fare_female = df[df['Sex'] == 'female']['Fare']

t_stat, p_value = stats.ttest_ind(fare_male, fare_female, equal_var=False)

print(f"Homens:    n={len(fare_male)}, media={fare_male.mean():.4f}, dp={fare_male.std():.4f}")
print(f"Mulheres:  n={len(fare_female)}, media={fare_female.mean():.4f}, dp={fare_female.std():.4f}")
print(f"Estatística t: {t_stat:.4f}")
print(f"p-valor: {p_value:.6f}")
print()

if p_value < alpha:
    print(f"p-valor ({p_value:.6f}) < alpha ({alpha}) -> Rejeitamos H0")
    print("Conclusão: Homens e mulheres pagaram tarifas médias significativamente diferentes.")
else:
    print(f"p-valor ({p_value:.6f}) >= alpha ({alpha}) -> Não rejeitamos H0")
    print("Conclusão: Não há evidência de que homens e mulheres pagaram tarifas diferentes.")
```

```python
# EXERCÍCIO 3: Passageiros da 1ª classe pagaram mais que da 3ª?
# H0: mu_1classe <= mu_3classe
# H1: mu_1classe > mu_3classe — teste unilateral (à direita)
# Usamos o teste t de Welch

alpha = 0.05

fare_p1 = df[df['Pclass'] == 1]['Fare']
fare_p3 = df[df['Pclass'] == 3]['Fare']

t_stat, p_value_two = stats.ttest_ind(fare_p1, fare_p3, equal_var=False)
p_value = p_value_two / 2  # unilateral à direita

print(f"1ª classe: n={len(fare_p1)}, media={fare_p1.mean():.4f}, dp={fare_p1.std():.4f}")
print(f"3ª classe: n={len(fare_p3)}, media={fare_p3.mean():.4f}, dp={fare_p3.std():.4f}")
print(f"Estatística t: {t_stat:.4f}")
print(f"p-valor (bilateral):  {p_value_two:.6f}")
print(f"p-valor (unilateral): {p_value:.6f}")
print()

if p_value < alpha and t_stat > 0:
    print(f"p-valor ({p_value:.6f}) < alpha ({alpha}) e t > 0 -> Rejeitamos H0")
    print("Conclusão: Passageiros da 1ª classe pagaram significativamente mais que os da 3ª classe.")
else:
    print(f"Não rejeitamos H0 -> Não há evidência suficiente de que a 1ª classe pagou mais.")
```

```python
# EXERCÍCIO 4: A tarifa média dos passageiros que sobreviveram é diferente de 50?
# H0: mu_sobreviventes = 50
# H1: mu_sobreviventes != 50 — teste bilateral

alpha = 0.05
mu0 = 50

fare_surv = df[df['Survived'] == 1]['Fare']

t_stat, p_value = stats.ttest_1samp(fare_surv, mu0)

print(f"Sobreviventes: n={len(fare_surv)}, media={fare_surv.mean():.4f}, dp={fare_surv.std():.4f}")
print(f"Estatística t: {t_stat:.4f}")
print(f"p-valor: {p_value:.6f}")
print()

if p_value < alpha:
    print(f"p-valor ({p_value:.6f}) < alpha ({alpha}) -> Rejeitamos H0")
    print("Conclusão: A tarifa média dos sobreviventes é significativamente diferente de 50.")
else:
    print(f"p-valor ({p_value:.6f}) >= alpha ({alpha}) -> Não rejeitamos H0")
    print("Conclusão: Não há evidência de que a tarifa média dos sobreviventes seja diferente de 50.")
```

```python
# EXERCÍCIO 5: Sobreviventes pagaram tarifas diferentes dos não-sobreviventes?
# H0: mu_sobreviventes = mu_nao_sobreviventes
# H1: mu_sobreviventes != mu_nao_sobreviventes — teste bilateral
# Usamos o teste t de Welch

alpha = 0.05

fare_surv = df[df['Survived'] == 1]['Fare']
fare_nonsurv = df[df['Survived'] == 0]['Fare']

t_stat, p_value = stats.ttest_ind(fare_surv, fare_nonsurv, equal_var=False)

print(f"Sobreviventes:     n={len(fare_surv)}, media={fare_surv.mean():.4f}, dp={fare_surv.std():.4f}")
print(f"Não-sobreviventes:  n={len(fare_nonsurv)}, media={fare_nonsurv.mean():.4f}, dp={fare_nonsurv.std():.4f}")
print(f"Estatística t: {t_stat:.4f}")
print(f"p-valor: {p_value:.6f}")
print()

if p_value < alpha:
    print(f"p-valor ({p_value:.6f}) < alpha ({alpha}) -> Rejeitamos H0")
    print("Conclusão: Sobreviventes e não-sobreviventes pagaram tarifas médias significativamente diferentes.")
else:
    print(f"p-valor ({p_value:.6f}) >= alpha ({alpha}) -> Não rejeitamos H0")
    print("Conclusão: Não há evidência de que sobreviventes e não-sobreviventes pagaram tarifas diferentes.")
```

```python
# EXERCÍCIO 6: Passageiros da 1ª classe pagaram MAIS que os da 2ª?
# H0: mu_1classe <= mu_2classe
# H1: mu_1classe > mu_2classe — teste unilateral (à direita)
# Usamos o teste t de Welch

alpha = 0.05

fare_p1 = df[df['Pclass'] == 1]['Fare']
fare_p2 = df[df['Pclass'] == 2]['Fare']

t_stat, p_value_two = stats.ttest_ind(fare_p1, fare_p2, equal_var=False)
p_value = p_value_two / 2  # unilateral à direita

print(f"1ª classe: n={len(fare_p1)}, media={fare_p1.mean():.4f}, dp={fare_p1.std():.4f}")
print(f"2ª classe: n={len(fare_p2)}, media={fare_p2.mean():.4f}, dp={fare_p2.std():.4f}")
print(f"Estatística t: {t_stat:.4f}")
print(f"p-valor (bilateral):  {p_value_two:.6f}")
print(f"p-valor (unilateral): {p_value:.6f}")
print()

if p_value < alpha and t_stat > 0:
    print(f"p-valor ({p_value:.6f}) < alpha ({alpha}) e t > 0 -> Rejeitamos H0")
    print("Conclusão: Passageiros da 1ª classe pagaram significativamente mais que os da 2ª classe.")
else:
    print(f"Não rejeitamos H0 -> Não há evidência suficiente de que a 1ª classe pagou mais que a 2ª.")
```
