### EXERCÍCIO 1: IMC

Calcule o IMC de uma pessoa:

IMC = peso / (altura ** 2)
- peso em kg
- altura em metros

Classifique: Normal (18.5-24.9)


```python
peso = input("Qual é seu peso? ")
altura = input("Qual é sua altura")

imc = float(peso)/(pow(float(altura),2))

if imc > 18.5 and imc <= 24.9:
    print(f"Seu {imc}, classificação: Normal")
else:
    print(f"Seu {imc}, classicação: Se cuida")
```

### EXERCÍCIO 2: MÉDIA DE NOTAS

Calcule a média de 5 notas e verifique se o aluno passou (≥6.0)

### EXERCÍCIO 4: ÁREA DO CÍRCULO

Área = π * raio² (use 3.14159 para π)

Tem outra maneira de utilizar o valor do número π?

```python
raio = float(input("Qual é o raio do círculo? "))

# Usando valor fixo para π
pi = 3.14159
area1 = pi * pow(raio, 2)
print(f"Área usando π fixo: {area1}")

# Usando biblioteca math para π
import math
area2 = math.pi * pow(raio, 2)
print(f"Área usando math.pi: {area2}")
```

### EXERCÍCIO 11: DESAFIO CIÊNCIA DE DADOS
### Análise completa de vendas de uma loja
### =====================================


DESAFIO: Analise os dados de vendas de uma loja e gere relatório completo

DADOS:
- vendas_jan: [1500, 2300, 1800, 2700, 3200, 1900, 2800, 2500, 3100, 2200]
- vendas_fev: [1700, 2100, 2600, 2900, 3400, 2000, 3000, 2700, 3300, 2400]
- produtos: ['Eletrônicos', 'Roupas', 'Alimentos', 'Móveis', 'Livros', 'Eletrônicos', 'Roupas', 'Alimentos', 'Móveis', 'Livros']

Tarefas:
1. Calcule total e média de cada mês
2. Identifique produto mais vendido (jan e fev)
3. Calcule crescimento % de jan→fev
4. Liste produtos acima da média geral
5. Crie classificação: 'Ótimo' (>3000), 'Bom' (2000-3000), 'Ruim' (<2000)

Use print formatados e operadores lógicos!

```python
#Dados
vendas_jan = [1500, 2300, 1800, 2700, 3200, 1900, 2800, 2500, 3100, 2200]
vendas_fev = [1700, 2100, 2600, 2900, 3400, 2000, 3000, 2700, 3300, 2400]
produtos = ['Eletrônicos', 'Roupas', 'Alimentos', 'Móveis', 'Livros', 'Eletrônicos', 'Roupas', 'Alimentos', 'Móveis', 'Livros']

#Variáveis
total_jan = sum(vendas_jan)
total_fev = sum(vendas_fev)
qtd_produtos = len(produtos)
qtd_vendas_jan = len(vendas_jan)
qtd_vendas_fev = len(vendas_fev)
media_jan = total_jan / len(vendas_jan)
media_fev = total_fev / len(vendas_fev)

#Info
print(f"Quantidade vendas jan {qtd_vendas_jan}")
print(f"Quantidade vendas fev {qtd_vendas_fev}")
print(f"Quantidade produtos {qtd_produtos}")

#Total e média de cada mês
print(f"Total de vendas em Janeiro {total_jan}, uma média de {media_jan}")
print(f"Total de vendas em Fevereiro {total_fev}, uma média de {media_fev}")

#Produto mais vendido (maior valor de venda)
indice_max_jan = vendas_jan.index(max(vendas_jan))
produto_mais_vendido_jan = produtos[indice_max_jan]
print(f"Produto mais vendido em Janeiro: {produto_mais_vendido_jan} (R$ {max(vendas_jan)})")

indice_max_fev = vendas_fev.index(max(vendas_fev))
produto_mais_vendido_fev = produtos[indice_max_fev]
print(f"Produto mais vendido em Fevereiro: {produto_mais_vendido_fev} (R$ {max(vendas_fev)})")

#Crescimento de vendas
crescimento = (total_fev - total_jan) / total_jan * 100
print(f"Crescimento de vendas: {crescimento:.2f}%")

#Produtos acima da média geral
media_geral = (total_jan + total_fev) / (qtd_vendas_jan + qtd_vendas_fev)
print(f"\nMédia geral: R$ {media_geral:.2f}")
print("\nProdutos acima da média geral:")
for i in range(len(vendas_jan)):
    if vendas_jan[i] > media_geral:
        print(f"- {produtos[i]} em Janeiro: R$ {vendas_jan[i]}")
    if vendas_fev[i] > media_geral:
        print(f"- {produtos[i]} em Fevereiro: R$ {vendas_fev[i]}")

#Classificação das vendas
print("\nClassificação das vendas:")
print("\nJANEIRO:")
for i in range(len(vendas_jan)):
    if vendas_jan[i] > 3000:
        classificacao = "Ótimo"
    elif vendas_jan[i] >= 2000:
        classificacao = "Bom"
    else:
        classificacao = "Ruim"
    print(f"{produtos[i]}: R$ {vendas_jan[i]} - {classificacao}")

print("\nFEVEREIRO:")
for i in range(len(vendas_fev)):
    if vendas_fev[i] > 3000:
        classificacao = "Ótimo"
    elif vendas_fev[i] >= 2000:
        classificacao = "Bom"
    else:
        classificacao = "Ruim"
    print(f"{produtos[i]}: R$ {vendas_fev[i]} - {classificacao}")

```
