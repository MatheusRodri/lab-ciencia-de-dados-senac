```python
# 1. VERIFICANDO O AMBIENTE
print("Bem-vindo à Aula 1!")
print(f"Python versão: {platform.python_version()}")
```

```python
# 2. NÚMEROS E OPERAÇÕES BÁSICAS
print("\n" + "="*50)

print("2. NÚMEROS E OPERAÇÕES")
print("="*50)

# Operações básicas
soma = 15 + 23
subtracao = 100 - 45
multiplicacao = 8 * 7
divisao = 20 / 4
potencia = 3 ** 4  # 3 elevado a 4
modulo = 25 % 7    # resto da divisão

print(f"Soma: {soma}")
print(f"Subtração: {subtracao}")
print(f"Multiplicação: {multiplicacao}")
print(f"Divisão: {divisao}")
print(f"Potência (3⁴): {potencia}")
print(f"Módulo (25%7): {modulo}")
```

```python
# Funções matemáticas úteis
print("\nFunções úteis:")
print(f"Valor absoluto: abs(-15) = {abs(-15)}")
print(f"Arredondar: round(3.14159, 2) = {round(3.14159, 2)}")
print(f"Potência: pow(2, 10) = {pow(2, 10)}")
```

```python
# 3. TIPOS DE DADOS
print("\n" + "="*50)
print("3. TIPOS DE DADOS")
print("="*50)

inteiro = 42
decimal = 3.14159
texto = "Ciência de Dados"
lista = [10, 20, 30, 40]

print(f"int: {inteiro} → tipo: {type(inteiro)}")
print(f"float: {decimal} → tipo: {type(decimal)}")
print(f"str: '{texto}' → tipo: {type(texto)}")
print(f"list: {lista} → tipo: {type(lista)}")
```

```python
# Conversões
print("Conversões:\n")
print(f"int(3.7) → {int(3.7)}")
print(f"float(10) → {float(10)}")
print(f"str(42) → '{str(42)}'")
```

```python
# 4. VARIÁVEIS - BOAS PRÁTICAS
print("\n" + "="*50)
print("4. VARIÁVEIS")
print("="*50)

# Nomes válidos (letras, números, _, não começar com número)
idade_aluno = 22
nota_prova = 8.5
nome_curso = "Sistemas de Informação"
notas_turma = [7.5, 8.0, 9.2, 6.8]

print(f"Idade: {idade_aluno}")
print(f"Nota: {nota_prova}")
print(f"Curso: {nome_curso}")
print(f"Média turma: {sum(notas_turma)/len(notas_turma):.2f}")
```

```python
# ERROS COMUNS (descomente para testar)
# 2nome = "inválido"  # SyntaxError
# meu-variavel = 10   # SyntaxError
# MinhaVar != minhaVar  # case-sensitive!
```

```python
# 5. STRINGS E FORMATAÇÃO
print("\n" + "="*50)
print("5. TRABALHANDO COM TEXTOS")
print("="*50)

cabecalho = "RELATÓRIO DE NOTAS"
aluno = "Aluno Alunos"
nota_final = 8.75

# Concatenação
print(cabecalho)
print(f"Aluno: {aluno}")
print(f"Nota final: {nota_final:.1f}")
```

```python
# Métodos de string
texto_minusculo = "Ciência de Dados".lower()
texto_maiusculo = "ciência de dados".upper()
print(f"Minúsculo: {texto_minusculo}")
print(f"Maiúsculo: {texto_maiusculo}")
```

```python
# 6. OPERAÇÕES COM LISTAS (PREVIEW CIÊNCIA DE DADOS)
print("\n" + "="*50)
print("6. LISTAS - BASE PARA DATAFRAMES")
print("="*50)

vendas_loja = [1500, 2300, 1800, 2700, 3200]
print(f"Vendas: {vendas_loja}")

# Estatísticas básicas
total_vendas = sum(vendas_loja)
media_vendas = total_vendas / len(vendas_loja)
venda_maxima = max(vendas_loja)

print(f"Total: R$ {total_vendas:,.2f}")
print(f"Média: R$ {media_vendas:.2f}")
print(f"Maior venda: R$ {venda_maxima:,.2f}")
```

```python
# 7. OPERADORES LÓGICOS
print("\n" + "="*50)
print("7. OPERADORES LÓGICOS")
print("="*50)

nota_aprovacao = 7.0
nota_aluno = 8.5
frequencia = 85

aprovado = (nota_aluno >= nota_aprovacao) and (frequencia >= 75)
print(f"Aprovado? {aprovado}")
```
